```python
# Pandas

# 1. importar pandas

import pandas as pd
```

```python
# 2. importar JSON

url = "https://www.datos.gov.co/resource/sdmr-tfmf.json"

df_json = pd.read_json(url)
print(df_json)
```

```python
# 3. Ver las primeras filas

print(df_json.head()) # ----> muestra las primeras 5 lineas
```

```python
# 4. Ver las ultimas filas

print(df_json.tail()) # ----> muestra las ultimas 5 lineas
```

```python
# 5. informacion del dataframe

print(df_json.info())
```

```python
# 6. estadistica basica

print(df_json.describe())
```

```python
# 7. ver nombres de columnas

print(df_json.columns)
```

```python
# 8. Seleccionar una columna

print(df_json["principio_activo1"])
```

```python
# 9. seleccionar varias columnas

print(df_json[["tipo_de_solicitud", "nombre_comercial_"]])
```

```python
# 10. Seleccionar una fila

print(df_json.iloc[2])
```

```python
# 11. Eliminar columnas

df = df_json.drop("forma_farmac_utica", axis=1)
print(df)
```

```python
# 12. ordenar datos

order = df_json.sort_values("concentraci_n_delmedicamento1")
print(order)
```

```python
# 13. ordenar descendiente

order_desc = df_json.sort_values("cantidad_solicitada", ascending=False)
print(order_desc)
```

```python
# 14. contar valores nulos

print(df_json.isnull().sum())
```

```python
# 15. llenar valores nulos

df_json["unidad_medida1"] = df_json["unidad_medida1"].fillna("Sin especificar")
print(df_json)
```

```python
# 17. eliminar valores nulos

df_limpio = df_json.dropna()
print(df_limpio)
```

```python
# 18. guardar en csv

df_json.to_csv("medicamentos_autorizados.csv", index=False)
print("Archivo guardado con éxito.")
```
